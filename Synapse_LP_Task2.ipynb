# Now let's start the fun part 😃

### Have you ever wondered what happens when you call **'regressor.fit(X_train, y_train)'** ?

### To understand what's hapenning in .fit method we will be implementing Linear Regression from scratch.

```python
class LinearRegression() :

    def __init__( self, learning_rate, iterations ) :
        # initialize the learning rate and iterations provides by the user
        self.learning_rate = learning_rate

        self.iterations = iterations

    # Function for model training
    def fit( self, X, Y ) :

        # no_of_training_examples, no_of_features
        self.m, self.n = X.shape

        # weight initialization
        self.W = np.zeros( self.n ) #set all initial weights as zero

        self.b = 0

        self.X = X

        self.Y = Y


        # gradient descent learning
        for i in range(self.iterations) : # complete the range function

            self.update_weights()

        return self

    # Helper function to update weights in gradient descent
    def update_weights(self) :

        Y_pred = (self.X).dot(self.W) + self.b

        # calculate gradients
        dW =  -2/self.m * (self.X.T).dot(self.Y - Y_pred)

        db = -2/self.m * np.sum(self.Y - Y_pred)

        # write the code to update the weights
        self.W = self.W - self.learning_rate * dW
        self.b = self.b - self.learning_rate * db

        return self


    def predict(self,X) :

        return X.dot(self.W) + self.b
```

```python
model = LinearRegression( iterations = 10000, learning_rate = 0.0015 )

model.fit(X_train, y_train)
```


## Question: What is this function actually doing? 🤔
#### it is using gradient descent and we take partial derivative of loss function to find the direction in which it decreases and we move in that direction until the solution converges or we hit max number of iterations
<br>

##### Hey, not bad! But can we further improve this? Remember our arbitrary choice for n_neighbours? Can we improve that?

### Choosing optimal 'k'

**Above, we chose n_neighbours to be equal to 1. Choosing a small value of K leads to unstable decision boundaries. <br>
We need to select n_neighbours by calculating the accuracy for every value of n from 1 to 60 and then choosing the one which gives the highest accuracy.**

##### **Plot a graph of K value vs Accuracy**

```python
best_k=0
best_accuracy=0
accuracy_vals=[]

for k in range(1, 61):
  model = KNeighborsClassifier(n_neighbors=k)
  model.fit(X_train, y_train)
  accuracy = model.score(X_test, y_test)
  accuracy_vals.append(accuracy)
  if accuracy > best_accuracy:
    best_accuracy = accuracy
    best_k = k
plt.plot(range(1, 61), accuracy_vals)
plt.xlabel('K')
plt.ylabel('Accuracy')
plt.show()
```

##### **Choose the best value of n_neighbours and give a reason why and also print the accuracy**

```python
print(f"the best accuracy we got is {best_accuracy} for k {best_k}")
```
